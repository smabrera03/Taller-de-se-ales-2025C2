```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
datos = np.loadtxt('mc.txt', delimiter = ',')

poblacion = datos[:, 0]
ganancias = datos[:, 1]

print(poblacion.shape)

poblacion = (poblacion - np.mean(poblacion))/np.std(poblacion)
ganancias = (ganancias - np.mean(ganancias))/np.std(ganancias)
```

```python
N = 100
a = np.linspace(start = -5, stop = 5, num = N)
b = np.linspace(start = -5, stop = 5, num = N)
error = np.zeros((N, N))
A, B = np.meshgrid(a, b)

for i in range(a.shape[0]):
    for j in range(b.shape[0]):
        error[i, j] = (1/N) * np.sum((ganancias - (a[i] * poblacion + b[j]))**2)


fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(A, B, error, cmap='viridis')

plt.show()
```

```python
X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

plt.show()
```
